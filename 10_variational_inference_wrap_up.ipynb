## Part 1: The big picture

The overall problem was always the same:

Given data $x$, we want to reason about hidden variables or parameters $z$ through the posterior
$$
p(z id x).
$$

Exact inference is often impossible, so we need approximations.

6. VAEs: VI for deep latent-variable models
7. SVI: minibatch optimization for large datasets

## Part 3: Core conceptual summary

If you remember only a few things, remember these:

### A. Bayesian inference is about uncertainty
We do not just want a point estimate; we want a distribution over plausible explanations.

### B. MCMC and VI solve the same problem differently
- MCMC approximates the posterior by sampling
- VI approximates the posterior by optimization

### C. The ELBO is central
Maximizing the ELBO means making the variational distribution closer to the true posterior.

### D. Simplicity vs expressiveness is the main trade-off
- Mean-field VI is simple and fast
- Flows and richer families are more flexible but cost more
